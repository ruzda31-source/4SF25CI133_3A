1. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Add a new column that is the product of the first two columns.

```python
import pandas as pd
import numpy as np

np.random.seed(7)
df = pd.DataFrame(
    np.random.randint(1, 20, size=(5, 3)),
    columns=["X", "Y", "Z"]
)

df["XY_Product"] = df["X"] * df["Y"]
print(df)
```

2. Create a Pandas DataFrame with 3 columns and 4 rows filled with random integers. Compute the row-wise and column-wise sum.

```python
import pandas as pd
import numpy as np

np.random.seed(21)
df = pd.DataFrame(
    np.random.randint(1, 15, size=(4, 3)),
    columns=["P", "Q", "R"]
)

print(df)
print("\nRow-wise sum:")
print(df.sum(axis=1))
print("\nColumn-wise sum:")
print(df.sum(axis=0))
```

3. Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Introduce some NaN values. Fill the NaN values with the mean of the respective columns.

```python
import pandas as pd
import numpy as np

np.random.seed(3)
df = pd.DataFrame(
    np.random.randint(10, 50, size=(5, 3)).astype(float),
    columns=["M1", "M2", "M3"]
)

# introduce NaN values
df.loc[1, "M1"] = np.nan
df.loc[3, "M2"] = np.nan
df.loc[4, "M3"] = np.nan
print("Before filling:")
print(df)

df = df.fillna(df.mean())
print("\nAfter filling with column means:")
print(df)
```

4. Create a Pandas DataFrame with 4 columns and 6 rows filled with random integers. Introduce some NaN values. Drop the rows with any NaN values.

```python
import pandas as pd
import numpy as np

np.random.seed(11)
df = pd.DataFrame(
    np.random.randint(100, 200, size=(6, 4)).astype(float),
    columns=["W", "X", "Y", "Z"]
)

df.iloc[0, 2] = np.nan
df.iloc[3, 0] = np.nan
df.iloc[4, 3] = np.nan
print("Original:")
print(df)

clean = df.dropna()
print("\nAfter dropping rows with NaN:")
print(clean)
```

5. Data Aggregation:
Create a Pandas DataFrame with 2 columns: 'Category' and 'Value'. Fill the 'Category' column with random categories ('A', 'B', 'C') and the 'Value' column with random integers. Group the DataFrame by 'Category' and compute the sum and mean of 'Value' for each category.

```python
import pandas as pd
import numpy as np

np.random.seed(5)
df = pd.DataFrame({
    "Category": np.random.choice(["A", "B", "C"], size=9),
    "Value": np.random.randint(10, 100, size=9)
})
print(df)

result = df.groupby("Category")["Value"].agg(["sum", "mean"])
print("\nSum and mean by category:")
print(result)
```

6. Create a Pandas DataFrame with 3 columns: 'Product', 'Category', and 'Sales'. Fill the DataFrame with random data. Group the DataFrame by 'Category' and compute the total sales for each category.

```python
import pandas as pd

df = pd.DataFrame({
    "Product":  ["Laptop", "Mouse", "Notebook", "Keyboard", "Pen", "Monitor", "Stapler"],
    "Category": ["Electronics", "Electronics", "Stationery", "Electronics", "Stationery", "Electronics", "Stationery"],
    "Sales":    [55000, 700, 120, 1500, 40, 12000, 250]
})
print(df)

total_sales = df.groupby("Category")["Sales"].sum()
print("\nTotal sales per category:")
print(total_sales)
```

7. Merging DataFrames:
Create two Pandas DataFrames with a common column. Merge the DataFrames using the common column.

```python
import pandas as pd

employees = pd.DataFrame({
    "Emp_ID": [101, 102, 103, 104],
    "Name":   ["Meera", "Arjun", "Fatima", "Dev"]
})

salaries = pd.DataFrame({
    "Emp_ID": [103, 101, 104, 102],      # different order on purpose
    "Salary": [52000, 48000, 61000, 45000]
})

merged = pd.merge(employees, salaries, on="Emp_ID")
print(merged)
```

8. MultiIndex DataFrame:
Create a Pandas DataFrame with a MultiIndex (hierarchical index). Perform some basic indexing and slicing operations on the MultiIndex DataFrame.

```python
import pandas as pd

index = pd.MultiIndex.from_tuples(
    [("Sem1", "Anil"), ("Sem1", "Bina"), ("Sem1", "Chetan"),
     ("Sem2", "Anil"), ("Sem2", "Bina"), ("Sem2", "Chetan")],
    names=["Semester", "Student"]
)
df = pd.DataFrame({"Marks": [72, 85, 64, 78, 90, 70]}, index=index)
print(df)

print("\nAll rows of Sem1:")
print(df.loc["Sem1"])

print("\nBina's marks in Sem2:")
print(df.loc[("Sem2", "Bina")])

print("\nAnil across all semesters:")
print(df.xs("Anil", level="Student"))

print("\nSlice from Sem1/Bina to Sem2/Anil:")
print(df.loc[("Sem1", "Bina"):("Sem2", "Anil")])
```

9. Create a Pandas DataFrame with MultiIndex consisting of 'Category' and 'SubCategory'. Fill the DataFrame with random data and compute the sum of values for each 'Category' and 'SubCategory'.

```python
import pandas as pd
import numpy as np

np.random.seed(9)
index = pd.MultiIndex.from_arrays(
    [["Fruit", "Fruit", "Fruit", "Fruit", "Veg", "Veg", "Veg", "Veg"],
     ["Apple", "Apple", "Mango", "Mango", "Carrot", "Carrot", "Onion", "Onion"]],
    names=["Category", "SubCategory"]
)
df = pd.DataFrame({"Value": np.random.randint(5, 50, size=8)}, index=index)
print(df)

print("\nSum for each Category and SubCategory:")
print(df.groupby(level=["Category", "SubCategory"]).sum())

print("\nSum for each Category:")
print(df.groupby(level="Category").sum())
```

10. Applying Functions:
Create a Pandas DataFrame with 3 columns and 5 rows filled with random integers. Apply a function that doubles the values of the DataFrame.

```python
import pandas as pd
import numpy as np

np.random.seed(13)
df = pd.DataFrame(
    np.random.randint(1, 10, size=(5, 3)),
    columns=["A", "B", "C"]
)
print("Original:")
print(df)

def double(col):
    return col * 2

doubled = df.apply(double)
print("\nAfter applying double():")
print(doubled)
```

11. Create a Pandas DataFrame with 3 columns and 6 rows filled with random integers. Apply a lambda function to create a new column that is the sum of the existing columns.

```python
import pandas as pd
import numpy as np

np.random.seed(17)
df = pd.DataFrame(
    np.random.randint(1, 30, size=(6, 3)),
    columns=["Maths", "Physics", "Chemistry"]
)

df["Total"] = df.apply(lambda row: row["Maths"] + row["Physics"] + row["Chemistry"], axis=1)
print(df)
```
